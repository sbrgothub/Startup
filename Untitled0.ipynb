What is considered a "success"

A startup is called successful if it makes a lot of money through:

1. M&A

Merger and Acquisition

This means:

A bigger company buys the startup

Example:

Facebook buying Instagram.

The founders get large money from the sale.

2. IPO

Initial Public Offering

This means:

The startup becomes a public company

Its shares are sold in the stock market.

Example:

Uber IPO.

Reasons for Failing

```python
closed=df.loc[df['status']=='closed']
```

```python
def success(x):
  if x=='acquired' or x=='ipo':
    return 'Success'
  elif x=='closed':
    return 'Fail'
  else:
    return 'Not Decided'

df['success_class']=df['status'].apply(success)
```

```python
success=df.loc[df['success_class']=='Success']
closed=df.loc[df['success_class']=='Fail']
```

```python
success_country={}
cnt_s=success['country_code'].unique()
cnt_f=closed['country_code'].unique()
for i in cnt_s:
  val_tot=df.loc[df['country_code']==i,'country_code'].value_counts()[0]
  val_success=success.loc[success['country_code']==i,'country_code'].value_counts()[0]
  success_country[i]=val_success/val_tot*100
```

State and region:

```python
success_state={}
cnt_s=success['state_code'].unique()

for i in cnt_s:
  val_tot=df.loc[df['state_code']==i,'state_code'].value_counts()[0]
  val_success=success.loc[success['state_code']==i,'state_code'].value_counts()[0]
  success_state[i]=val_success/val_tot*100
```

```python
cnt_f=closed['state_code'].unique()
failure_state={}
for i in cnt_f:
  val_tot=df.loc[df['state_code']==i,'state_code'].value_counts()[0]
  val_closed=closed.loc[closed['state_code']==i,'state_code'].value_counts()[0]
  failure_state[i]=val_closed/val_tot*100
```

```python
def state_score(x):
  if x in success_state.keys():
    if x in failure_state.keys():
      return success_state[x]/(success_state[x]+failure_state[x])
    else:
      return 1
  else:
    return 0
df['state_score']=df['state_code'].apply(state_score)
```

```python
pd.cut(df['state_score'],bins=5).value_counts()
def scorer(x):
  if x>0.8:
    return 'A'
  elif x>0.6:
    return 'B'
  elif x>0.4:
    return 'C'
  elif x>0.2:
    return 'D'
  else:
    return 'E'
df['state_code']=df['state_score'].apply(scorer)
```

Region

```python
success_region={}
cnt_s=success['region'].unique()

for i in cnt_s:
  val_tot=df.loc[df['region']==i,'region'].value_counts()[0]
  val_success=success.loc[success['region']==i,'region'].value_counts()[0]
  success_region[i]=val_success/val_tot*100
```

```python
cnt_f=closed['region'].unique()
failure_region={}
for i in cnt_f:
  val_tot=df.loc[df['region']==i,'region'].value_counts()[0]
  val_closed=closed.loc[closed['region']==i,'region'].value_counts()[0]
  failure_region[i]=val_closed/val_tot*100
```

```python
def region_score(x):
  if x in success_region.keys():
    if x in failure_region.keys():
      return success_region[x]/(success_region[x]+failure_region[x])
    else:
      return 1
  else:
    return 0
df['region_score']=df['region'].apply(region_score)
```
